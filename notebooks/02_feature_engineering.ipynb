```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import AllChem
import numpy as np
df = pd.read_csv('../data/egfr_pIC50_data.csv')
def smiles_to_fingerprint(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles) 
        fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius=2, nBits=2048)
        return np.array(fp)
    except:
        return None
df['Fingerprint'] = df['canonical_smiles'].apply(smiles_to_fingerprint)
df = df.dropna(subset=['Fingerprint'])
print(f"vectorized {len(df)} molecules")
df.head(3)
```

```python
df['Fingerprint'] = df['Fingerprint'].apply(lambda x: ",".join(map(str, x)))
df.to_csv('../data/egfr_with_fingerprints.csv', index=False)
```
