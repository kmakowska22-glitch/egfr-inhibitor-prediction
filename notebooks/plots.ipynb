```python
import os
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs("results/figures", exist_ok=True)

df = pd.read_csv("egfr_pIC50_data.csv")

plt.figure(figsize=(9,5))

plt.hist(
    df["pIC50"].dropna(),
    bins=25
)

plt.xlabel("pIC50")
plt.ylabel("Number of Compounds")
plt.title("Distribution of EGFR Inhibitor Activity")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    "results/figures/pic50_distribution.png",
    dpi=300
)

plt.show()


results = pd.read_csv("final_model_results.csv")

results = results[
    (results["pIC50"] > 1) &
    (results["pIC50"] < 12)
]

actual_col = "pIC50"
pred_col = "predicted_pIC50"

min_val = min(
    results[actual_col].min(),
    results[pred_col].min()
)

max_val = max(
    results[actual_col].max(),
    results[pred_col].max()
)

plt.figure(figsize=(7,7))

plt.scatter(
    results[actual_col],
    results[pred_col],
    alpha=0.4,
    s=15
)

plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    "--",
    linewidth=2
)

plt.xlabel("Actual pIC50")
plt.ylabel("Predicted pIC50")
plt.title("Predicted vs Actual pIC50")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    "results/figures/predicted_vs_actual.png",
    dpi=300
)

plt.show()

print("Plots saved successfully.")
```
